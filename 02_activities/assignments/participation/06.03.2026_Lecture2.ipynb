## How do we choose the right visualization? (Activity)

### Visualization #1: US Gun Killings in 2018

#### What information can we learn from this visualization?

- Number of gun killings, age of victim, estimated difference in lifespan
- Some data are summarized from existing data and some are predicted data

#### Is this an example of objective, neutral data visualization? Why or why not?

- Not very objective in terms of prediction, not neutral
- Descriptions like "if they didn't die from the gun, they would be killed by mental disorder/cancer"
- Use vocabulary that is intended to invoke emotional response (loss)
- Whether it is effective depends on final purpose of the plot

### Visualization #2: Washington Post Active Shooters

#### What information can we learn from this visualization?

- Number of active shooter incidents annually
- Increase of active shooters over time

#### Is this an example of objective, neutral data visualization? Why or why not?

- More objective compared to other 
- What is "active" ? Depends on how FBI defines; not defined on the graph
- No vocabulary intended to strike emotional response
- Impact is not scoped, needs more clarity (Is it for a state or entire country?)

## Can data visualization be neutral?

Short answer, no. How to design data visualization is a choice that will always impact our interpretation. But, can be factual without being neutral

## What do we want our data visualization to do?

Intended purpose (persuade, compare, evaluate, explore) \
Intended audience (Age, education, expertise, accessibility) \
Intended medium (print, web, poster, presentation) 

A good goal is to satisfy: “to find one of the many possible good solutions rather than
one of the even larger number of bad ones”

## Gestalt Principles

Proximity (Close together = in a group) \
Similarity (Similar objects = grouped eg. shape) \ 
Continuity (Aligned objects = grouped) \ 
Closure (Open structures = closed) \

Consider the cognitive load / work required to take into new information


## Factual Bias

Adherance to four conventions of data visualizations (Sociologists):

- 2D image
- Clean layout
- Geometric shapes and lines
- Inclusion of data sources at bottom of image

# Exploring matplotlib (activity)

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')

sns.set_theme(style="darkgrid")
sns.kdeplot(df['sepal_width'], fill=True)
plt.show()
```

## Aesthetic

The plot is very visually appealing. The grid lines make it very easy to interpret exact values on the axes. The filled in histogram makes it very easy to see the distribution of sepal widths.

## Substantive

This plot is relatively objective and straightforward. The code indicates that this plot is pulling data as expected to generate the plot. 

## Perceptual

This plot does not require a large cognitive load as it is relatively consise and follows a familiar chart type.
